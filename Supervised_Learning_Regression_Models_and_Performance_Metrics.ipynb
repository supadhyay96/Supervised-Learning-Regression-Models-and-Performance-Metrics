    **Q1.  What is Simple Linear Regression (SLR)? Explain its purpose.**

        -> # **Simple Linear Regression (SLR) and Its Purpose**

    **Simple Linear Regression (SLR)** is a statistical and supervised machine learning technique used to identify the relationship between two variables and predict the value of one variable based on another.

    It involves two variables:

    - **Independent Variable (X):** The variable used to make predictions.
    - **Dependent Variable (Y):** The variable whose value is predicted.

    ### **1. Mathematical Equation**

    The general equation of Simple Linear Regression is:

    \\[ Y = a + bX \\]

    Where:

    - **Y:** Predicted value of the dependent variable.
    - **X:** Independent variable.
    - **a:** Intercept (value of Y when X = 0).
    - **b:** Slope (change in Y for a one-unit change in X).

    ### **2. Example**

    Suppose we want to predict students' exam marks based on study hours.

    | Study Hours (X) | Marks (Y) |
    | --------------- | --------- |
    | 1               | 30        |
    | 2               | 40        |
    | 3               | 50        |
    | 4               | 60        |
    | 5               | 70        |

    Regression equation:

    \\[ Y = 20 + 10X \\]

    If a student studies for 6 hours:

    \\[ Y = 20 + 10(6) = 80 \\]

    Therefore, the predicted marks are **80**.

    ### **3. Graphical Representation**

    SLR represents the relationship between two variables using a straight line called the regression line.

    ### **4. Purpose of Simple Linear Regression**

    1. **Prediction:** Predicts unknown values using available data.
    2. **Relationship Analysis:** Identifies the relationship between two variables.
    3. **Trend Identification:** Determines whether variables have a positive or negative relationship.
    4. **Forecasting:** Estimates future sales, profits and demand.
    5. **Decision Making:** Helps businesses make data-driven decisions.
    6. **Impact Analysis:** Measures how changes in one variable affect another.

    ### **5. Applications**

    - **Education:** Predicting marks based on study hours.

    ### **6. Advantages and Limitations**

    **Advantages:**

    - Easy to understand and implement.
    - Requires less computational power.
    - Useful for prediction and forecasting.
    - Helps identify relationships between variables.

    **Limitations:**

    - Uses only one independent variable.
    - Assumes a linear relationship.
    - Sensitive to outliers.

    - Does not establish cause-and-effect relationships.

    **Q2. What are the key assumptions of Simple Linear Regression?**

        -> # **Key Assumptions of Simple Linear Regression (SLR)**

    Simple Linear Regression is based on certain assumptions to ensure that the model produces reliable and accurate results. The key assumptions are:

    1\. **Linearity:** There must be a linear relationship between the independent variable (X) and dependent variable (Y).

    2\. **Independence:** Observations and error terms should be independent of each other.

    3\. **Homoscedasticity:** The variance of error terms (residuals) should remain constant across all values of X.

    4\. **Normality of Errors:** The residuals (differences between actual and predicted values) should follow an approximately normal distribution.

    5\. **No Significant Outliers:** The dataset should not contain extreme outliers that can strongly affect the regression line.

    6\. **Zero Mean of Errors:** The average value of the error terms should be approximately zero.

    ### **Conclusion**

    These assumptions help ensure that Simple Linear Regression provides reliable predictions and meaningful statistical results. If these assumptions are violated, the model's accuracy and reliability may decrease.

            -> ## **Mathematical Equation of Simple Linear Regression (SLR)**

        The mathematical equation of a Simple Linear Regression model is:

        \\[ Y = \beta_0 + \beta_1X + \epsilon \\]

        **Explanation of each term:**

        1. **Y (Dependent Variable):** The variable whose value we want to predict.
        2. **X (Independent Variable):** The variable used to predict the dependent variable.
        3. **β₀ (Intercept):** The expected value of Y when X is zero.
        4. **β₁ (Slope/Coefficient):** Represents the change in Y for every one-unit increase in X.
        5. **ε (Error Term):** Represents the difference between the actual value and the value predicted by the model.

        **Example:**

        Suppose the regression equation is:

           Y = 20 + 10X

        If X = 5, then:

         Y = 20 + 10(5) = 70

        Therefore, the predicted value of Y is **70**.

        **Conclusion:** Simple Linear Regression uses a straight-line equation to describe the relationship between two variables and predict the dependent variable based on the independent variable.

    ## Method of Least Squares in Linear Regression

    The **Method of Least Squares** is a statistical technique used in linear regression to find the best-fitting straight line for a given dataset. It minimizes the sum of squared differences between actual values and predicted values. These differences are called residuals or errors.

    ### 1. Mathematical Formula

    SSE = Σ (Yi - Ŷi)²

    Where:

    - Yi = Actual value
    - Ŷi = Predicted value
    - n = Total number of observations
    - SSE = Sum of Squared Errors

    ### 2. Working Principle

    1. Collect data containing independent (X) and dependent (Y) variables.
    2. Fit a straight regression line to the data.
    3. Calculate predicted values.
    4. Find the difference between actual and predicted values.
    5. Square the errors and calculate their sum.
    6. Choose the line with the minimum sum of squared errors.

    ### 3. Purpose

    - Finds the best-fitting regression line.
    - Minimizes prediction errors.
    - Improves prediction reliability.
    - Helps understand the relationship between two variables.

          -> ## **1. What is Logistic Regression?**

      **Logistic Regression** is a supervised machine learning algorithm used for classification problems. It predicts the probability of an outcome belonging to a particular category, usually binary classes such as Yes/No, True/False or 0/1.

      It uses the **Sigmoid Function** to convert predicted values into probabilities between 0 and 1.

      ### **Mathematical Equation**

      P(Y=1) = 1 / (1 + e^(-z))

      Where:

      - P(Y=1) = Probability of the positive class.
      - e = Mathematical constant (approximately 2.718).
      - z = Linear combination of input variables.

      ### **Example**

      An email classification system uses Logistic Regression to identify whether an email is Spam or Not Spam.

      - Probability ≥ 0.5: Spam (1).
      - Probability < 0.5: Not Spam (0).

        **Q7.  Name and briefly describe three common evaluation metrics for regression models.**

            -> ## **Three Common Evaluation Metrics for Regression Models**

        Regression evaluation metrics are used to measure how accurately a regression model predicts numerical values. Three common metrics are:

        ### **1. Mean Absolute Error (MAE)**

        MAE measures the average absolute difference between actual and predicted values. A lower MAE indicates better model performance.

        ### **2. Mean Squared Error (MSE)**

        MSE calculates the average of the squared differences between actual and predicted values. It gives more penalty to larger errors. A lower MSE indicates better performance.

        ### **3. R-Squared (R² Score)**

        R² measures how well the independent variables explain the variation in the dependent variable. Its value generally ranges from 0 to 1, where a value closer to 1 indicates a better model fit.

        ### **Conclusion**

        MAE, MSE and R² are commonly used regression evaluation metrics. They help measure prediction errors and determine how well a regression model fits the given data.

      **Q8.  What is the purpose of the R-squared metric in regression analysis?**

          -> ## Purpose of R-Squared (R²) in Regression Analysis

      R-Squared (R²) is an evaluation metric used in regression analysis to measure how well a regression model explains the variation in the dependent variable using the independent variable(s).

      ### 1. Formula

      R² = 1 - (SSres / SStot)

      Where:

      - SSres = Sum of Squared Residuals (prediction errors).
      - SStot = Total Sum of Squares (total variation in actual values).

      ### 2. Purpose of R-Squared

      1. Measures how well the regression model fits the data.
      2. Indicates the percentage of variation explained by the model.
      3. Helps compare different regression models.
      4. Evaluates the explanatory power of a model.

      ### 3. Example

      If R² = 0.85, it means the model explains 85% of the variation in the dependent variable.

      ### 4. Conclusion

      R-Squared is used to evaluate the goodness of fit of a regression model. A value closer to 1 generally indicates a better fit, while a value closer to 0 indicates that the model explains less variation.
